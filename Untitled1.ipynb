```python
import torch
from torch.nn import *
def Sim(r, set, get, envk=.99, diffk=.2):
    torch.manual_seed(0)
    set = *torch.randint(r, (2,set)),
    get = *torch.randint(r, (2,get)),
    def Sim(x, n):
        a = torch.zeros(r,r)
        for n in range(n):
            a = a * envk
            _ = functional.pad(a[None], (1,1,1,1), 'replicate')[0]
            a = a + (_[:-2,1:-1] + _[2:,1:-1] - 4*a + _[1:-1,:-2] + _[1:-1,2:]) * diffk
            a = a.index_put(set, x)
        return a[get]
    return Sim
sim = Sim(8, 47, 34)
x = (torch.randn(47)+2).requires_grad_()
O = torch.optim.Adam([x], .01, eps=1e-15)
z = torch.tensor(0.)
for _ in range(100):
    for _ in range(1000):
        y = sim(x, 100)
        l = functional.mse_loss(y, z)
        if l < 1e-1:
            break
        l.backward()
        utils.clip_grad_norm_([x], 1)
        O.step()
        O.zero_grad()
    else:
        print('x', l.data)
        continue
    print('x', l.data)
    break
```
