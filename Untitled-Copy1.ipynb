```python
import torch
import numpy as np
from threading import Thread
from time import time, sleep
from torch.nn import *
import matplotlib.pyplot as plt
from IPython.display import clear_output
def add(a, b):
    return a if b is None else a+b
def dot(a, b):
    return None if a is None else a@b
def Sim(_=8, n=1, set=8, get=8, envk=.99, diffk=.2):
    set = torch.arange(n)[:,None], *torch.randint(1, _-1, (2,set))
    get = set[0], *torch.randint(1, _-1, (2,get))
    _ = torch.zeros(n,_,_)
    def Sim(setv):
        if setv is None:
            _[:] = 0
        else:
            _[:] *= envk
            _[set] = setv
            _[:,0] = _[:,1]
            _[:,-1] = _[:,-2]
            _[:,:,0] = _[:,:,1]
            _[:,:,-1] = _[:,:,-2]
            _[:,1:-1,1:-1] += (_[:,:-2,1:-1]+_[:,2:,1:-1]-4*_[:,1:-1,1:-1]+_[:,1:-1,:-2]+_[:,1:-1,2:])*diffk
            return _[get]
    return Sim
class Rnn(Linear):
    def __init__(s, i, o, h):
        super().__init__(i+o+h, h)
        s._ = Linear(h, o)
    def forward(s, i, o=None, h=None, l=-1):
        I,O,H = s.weight.mT.split([i.size(-1),s._.out_features,s.out_features])
        h = [h]
        if o is None or o.ndim<3:
            o = [o]
            for _ in functional.linear(i, I.mT, s.bias).broadcast_to(l,-1,-1):
                h += functional.silu(add(add(_, dot(o[-1], O)), dot(h[-1], H))),
                o += s._(h[-1]),
            return torch.stack(o[1:])
        for _ in add(functional.linear(i, I.mT, s.bias), dot(o, O)).broadcast_to(l,-1,-1):
            h += functional.silu(add(_, dot(h[-1], H))),
        return s._(torch.stack(h[1:]))
def OvershotLR(o, k=.5, w=1, K=1.005, W=1):
    N = 0
    L = 9e9
    def OvershotLR(l):
        nonlocal N,L
        l.backward()
        utils.clip_grad_norm_(o.param_groups[0]['params'], 1.)
        N = min(N,1)-1 if l>L else max(N,-1)+1
        o.param_groups[0]['lr'] = r = o.param_groups[0]['lr']*(k*w**N if l>L else K*W**N)
        o.step()
        o.zero_grad()
        L = l
        return l.data, r
    return OvershotLR
def Plot(**_):
    A = []
    def Plot(a=None):
        if a is None:
            plt.plot(A, **_)
        else:
            A.append(a)
    return Plot
```
