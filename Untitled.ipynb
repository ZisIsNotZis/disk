```python
import torch
from torch.nn import *
def Sim(_, n, set, get, envk=.99, setk=1., diffk=.2):
    set = torch.arange(n)[:,None], *torch.randint(1, _-1, (2,set))
    get = set[0], *torch.randint(1, _-1, (2,get))
    _ = torch.full((n,_,_), env)
    def Sim(setv):
        if setv is None:
            _[:] = 0
        else:
            _ *= envk
            _[set] += (setv-_[set])*setk
            _[:,0] = _[:,1]
            _[:,-1] = _[:,-2]
            _[:,:,0] = _[:,:,1]
            _[:,:,-1] = _[:,:,-2]
            _[:,1:-1,1:-1] += (_[:,:-2,1:-1]+_[:,2:,1:-1]-4*_[:,1:-1,1:-1]+_[:,1:-1,:-2]+_[:,1:-1,2:])*diffk
            return _[get]
    return Sim
def Scale(N=10):
    m,v,n = torch.tensor((0,1e-9,0))
    return lambda _:n.add_(1)and(_-m.add_(n<=N and _.data.mean())/min(n,N))/(v.add_(n<=N and _.data.var())/min(n,N))**.5,lambda _:(_*(v/min(n,N))**.5)+m/min(n,N)
```

```python
def step(a, env, set, setp, envk=1e-4, diffk=.2, setk=1.):
    if envk:
        a += (env-a)*envk
    a += (a.roll(1,0)+a.roll(-1,0)-4*a+a.roll(1,1)+a.roll(-1,1))*diffk
    a[set] += (setp-a[set])*setk
    return a
env = 23.
tgt = torch.tensor(200.)
set = *torch.randint(8,(2,8)),
get = *torch.randint(8,(2,8)),
setp = torch.rand(8,requires_grad=True)
o = torch.optim.Adam((setp,), 10, eps=1e-15)
L = []
R = []
l = 9e9
n = 0
for _ in range(100):
    a = torch.full((8,8), env)
    for _ in range(128):
        a = step(a, env, set, setp)
    l_ = functional.mse_loss(a[get], tgt)
    if l_ > l:
        n += 1
        o.param_groups[0]['lr'] *= .9**n
    else:
        n = 0
        o.param_groups[0]['lr'] *= 1.01
    l = l_
    l.backward()
    o.step()
    o.zero_grad()
    L.append(l.data)
    R.append(o.param_groups[0]['lr'])
plt.plot(L)
plt.yscale('log')
plt.grid()
plt.twinx().plot(R, color='orange')
plt.yscale('log')
print(setp)
```

```python
import torch
import matplotlib.pyplot as plt
from torch.nn import *
from IPython.display import clear_output
def step(a, env, set, setp, envk=1e-3, diffk=.2, setk=1.):
    if envk:
        a += (env-a)*envk
    a += (a.roll(1,-2)+a.roll(-1,-2)-4*a+a.roll(1,-1)+a.roll(-1,-1))*diffk
    a[set] += (setp-a[set])*setk
    return a
env = 23.
tgt = torch.tensor(200.)
```

```python
class Rnn(Linear):
    def __init__(s, i, o, h):
        super().__init__(i+o+h, h)
        s.h = Parameter(torch.zeros(h))
        s._ = Linear(h, o)
    def forward(s, _, o, h=None, n=-1):
        h = [s.h if h is None else h]
        if o.ndim>2:
            for _ in functional.linear(_,s.weight[:,:_.size(-1)],s.bias).broadcast_to(n,*(-1,)*(_.ndim-1))+o@s.weight.mT[_.size(-1):-len(s.h)]:
                h += functional.silu(_+h[-1]@s.weight.mT[-len(s.h):]),
            o = s._(torch.stack(h[1:]))
        else:
            o = [o]
            for _ in functional.linear(_,s.weight[:,:_.size(-1)],s.bias).broadcast_to(n,*(-1,)*(_.ndim-1)):
                h += functional.silu(_+o[-1]@s.weight.mT[-len(s.h)-o[-1].size(-1):-len(s.h)]+h[-1]@s.weight.mT[-len(s.h):]),
                o += s._(h[-1]),
            o = torch.stack(o[1:])
        return o, h[-1]
set = *torch.randint(8,(2,8)),
get = *torch.randint(8,(2,8)),
setv = torch.full((8,), env)
m = Rnn(8, 8, 16)
o = torch.optim.Adam(m.parameters(), .02)
o2 = torch.optim.Adam((setv,), .2)
scale,inv = Scale()
a = torch.full((8,8), env)
X = torch.empty(10001,1,8)
T = torch.empty(10001,1,8)
T[0] = env
L,R,L2,R2 = [],[],[],[]
for i in range(10000):
    x = setv + (10000-i)/100*torch.randn(8)
    X[i] = scale(x)
    a = step(a, env, set, x)
    T[i+1] = scale(a[get])
    if not(i+1)%100:
        l = 9e8
        for _ in range(100):
            l_ = functional.mse_loss(m(X[:i+1], T[:i+1])[0], T[1:i+2])
            if l_ > l:
                o.param_groups[0]['lr'] *= .999
            l = l_
            l.backward()
            o.step()
            o.zero_grad()
            L.append(l.data)
            R.append(o.param_groups[0]['lr'])
        m.requires_grad_(False)
        setv.requires_grad_()
        l = 9e8
        for _ in range(1000):
            l_ = functional.mse_loss(m(scale(setv)[None,None], T[i+1], h.data, 256)[0][-1], scale(tgt))
            if l_ > l:
                o2.param_groups[0]['lr'] *= .999
            l = l_
            l.backward()
            o2.step()
            o2.zero_grad()
            L2.append(l.data)
            R2.append(o2.param_groups[0]['lr'])
        setv.requires_grad_(False)
        m.requires_grad_()
        plt.subplot(211).plot(L)
        plt.yscale('log')
        plt.twinx().plot(R, c='red')
        plt.yscale('log')
        plt.subplot(212).plot(L2)
        plt.yscale('log')
        plt.twinx().plot(R2, c='red')
        plt.yscale('log')
        clear_output(True)
        plt.show()
```
